ITERATING (BREAKING OF ARRAY OR LIST)

```python
import numpy as np
var = np.array([1,2,3,4,5,6])
for i in var :
    print(i)
```

```python
var2 =np.array([[[1,2,3,4],[1,2,3,4]]])
print(var2)
print(var2.ndim)
print()
for i in np.nditer(var2, flags=['buffered'],op_dtypes=["S"]):
    print(i)
```

```python
var2 =np.array([[[1,2,3,4],[1,2,3,4]]])
print(var2)
print(var2.ndim)
print()
for i,d in np.ndenumerate(var2) :
    print(i,d)
```

JOINING AND SPLIT ARRAY

```python
import numpy as np
var = np.array([1,2,3,4])
var2= np.array([6,7,8,9])
ar = np.concatenate((var,var2))
print(ar)
```

```python
import numpy as np
var = np.array([[1,2],[3,4]])
var2= np.array([[6,7],[8,9]])
ar = np.concatenate((var,var2),axis=0)
print(var)
print(var2)
print(ar)
```

```python
import numpy as np
var = np.array([[1,2],[3,4]])
var2= np.array([[6,7],[8,9]])
ar = np.stack((var,var2),axis=0)
ar1 = np.hstack((var,var2))
ar2= np.dstack((var,var2))
ar3 = np.vstack((var,var2))

print(var)
print(var2)
print(ar)
print(ar1)
print(ar2)
print(ar3)
```
